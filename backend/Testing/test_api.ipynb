```python
import requests
```

```python
BASE_URL = "http://127.0.0.1:5000"
```

```python
register_payload = {
    "emailid": "cuncky@gmail.com",
    "username": "cuncky.jackson",
    "firstname": "Test",
    "lastname": "User",
    "password": "test1234",
    "confirm_password": "test1234",
    "age": 25,
    "gender": "Male",
    "industry": "Tech",
    "profession": "Engineer"
}

resp = requests.post(f"{BASE_URL}/api/register", json=register_payload)
print("Register Response:", resp.status_code, resp.json())
```

```python
login_payload = {
    "emailid": "cuncky@gmail.com",
    "password": "test1234"
}

resp = requests.post(f"{BASE_URL}/api/login", json=login_payload)
print("Login Response:", resp.status_code, resp.json())

user_id = resp.json().get("user_id")
```

```python
phq9_payload = {
    "user_id": user_id,
    "responses": {
        "1": 1, "2": 2, "3": 0, "4": 1, "5": 2,
        "6": 0, "7": 1, "8": 2, "9": 1
    },
    "totalScore": 9,
    "doctors_notes": "Sample doctor notes",
    "patients_notes": "Sample patient notes"
}

resp = requests.post(f"{BASE_URL}/api/phq9", json=phq9_payload)
print("PHQ-9 Response:", resp.status_code, resp.json())
```
